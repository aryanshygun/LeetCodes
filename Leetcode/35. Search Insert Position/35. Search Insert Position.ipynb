```python
from typing import List
```

```python
class Solution:
    def searchInsert(self, nums: List[int], target: int) -> int:
        if target in nums:
            return nums.index(target)
        else:
            count = 0 
            for i in nums:
                if i < target:
                    count += 1
            return count
```

```python
class Solution:
    def searchInsert(self, nums: List[int], target: int) -> int:
        if target in nums:
            return nums.index(target)
        count = 0 
        while count < len(nums):
            if nums[count] >= target:
                break
            count += 1
        return count
```
