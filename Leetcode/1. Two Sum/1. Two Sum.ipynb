```python
from typing import List
```

```python
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        i = 0
        while i <len(nums):
            j = i + 1
            while j < len(nums):
                if nums[i] + nums[j] == target:
                    return [i, j]
                j += 1
            i += 1
```

```python
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        xdict = {}
        for i, value in enumerate(nums):
            y = target - value
            if y in xdict:
                return [xdict[y], i]
            xdict[value] = i
```
